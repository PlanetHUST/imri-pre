```python
import numpy as np
from src.physical_process.constants import G, msun, c, Rsun, AU, yr
```

```python
M_BH_g=1e4*msun
m_rem_g=1e2*msun
a = AU

r_ISCO = 6*G*M_BH_g/c**2

print(f'ISCO radius is {r_ISCO:.3e} cm = {r_ISCO/Rsun:.3e} rsun')

red_shift = 10
M_tot = M_BH_g+m_rem_g
M_tot_rs = M_tot*(1+red_shift)

f_ISCO = c**3/(np.pi*6**1.5*G* M_tot_rs)

print(f'ISCO frequency is {f_ISCO:.3e} Hz')

time_gw_s = 5/256*c**5*a**4/(G**3*M_BH_g*m_rem_g*M_tot)
time_gw_yr = time_gw_s / yr

print(f'GW inspial time is {time_gw_yr:.3e} yr')
```

```python
h = 0.01
alpha = 0.01
Sigma_cgs = 1e5

rg_cm = G*M_BH_g/c**2

r_cm = 1e5 * rg_cm

Omega_cgs = np.sqrt(G*M_BH_g/r_cm**3)

time_type_i_s = M_BH_g**2*h**2/(m_rem_g*Sigma_cgs*r_cm**2*Omega_cgs)
time_type_i_yr=time_type_i_s/yr

print(f'type i migration timescale for initial radius {r_cm/rg_cm:.3e} rg is {time_type_i_yr:.3e} yr')

time_type_ii_s = (alpha*h**2*Omega_cgs)**-1
time_type_ii_yr=time_type_ii_s/yr

print(f'type ii migration timescale for initial radius {r_cm/rg_cm:.3e} rg is {time_type_ii_yr:.3e} yr')
```
